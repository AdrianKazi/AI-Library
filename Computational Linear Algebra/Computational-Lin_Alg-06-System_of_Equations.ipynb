# System of Equations

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
import numpy as np
import matplotlib.pyplot as plt
from IPython import display
import time

# coefficients of equations
# ay = bx + c

eq1o = np.array([1., 2., 1.])
eq2o = np.array([2., 1., 3.])

x = np.array([-3, 3])

for i in range(10):

    # clear plot
    # plt.cla()

    # randomly update equations
    eq1 = eq1o + np.random.randn(1) * eq2o
    eq2 = eq2o + np.random.randn(1) * eq1o

    # ay = bx + c  -> y = (bx+c)/a - we find y
    y1 = (eq1[1] * x + eq1[2]) / eq1[0]
    y2 = (eq2[1] * x + eq2[2]) / eq2[0]

    # plot lines
    plt.plot(x, y1, label='eq1')
    plt.plot(x, y2, label='eq2')

    plt.axis([-3, 3, -3, 6])
    plt.legend()
    plt.grid(True)

    # refresh display
    display.clear_output(wait=True) # without that new iter woutl add new plot under prev one
    display.display(plt.gcf())

    time.sleep(1)

plt.close()
```

3d

```python
eq1o = [1,2,3,-1]
eq2o = [2,1,3,3]

# az = bx + cy + d
# y = - (bx - az + d) / c
# z = (bx + cy + d) / a

# set up for 3D plot
fig = plt.figure()
ax = fig.add_subplot(projection = '3d')

for i in range(0,10):
  # ax.cla()

  # here we add to eq1o just the linear combination of eq2o and vice versa
  eq1 = np.add(eq2o, np.random.randn(1)*eq1o)
  eq2 = np.add(eq1o, np.random.randn(1)*eq2o)

  # plot new lines (solutions) for eq1
  y = ([eq1[1]*-3, eq1[1]*3] + eq1[3]) / eq1[0]
  z = ([eq1[2]*-3, eq1[2]*3] + eq1[3]) / eq1[0]
  ax.plot([-3,3], y, z)

  # plot new lines (solutions) for eq2
  y = ([eq2[1]*-3, eq2[1]*3] + eq2[3]) / eq2[0]
  z = ([eq2[2]*-3, eq2[2]*3] + eq2[3]) / eq2[0]
  ax.plot([-3,3], y, z)

  # axis limits
  ax.set_xlim3d(-3,6)
  ax.set_ylim3d(-3,6)
  ax.set_zlim3d(-1,10)

  display.clear_output(wait = True)
  display.display(plt.gcf())
  time.sleep(1)
```

# Gaussian Elimination

6-step plan for solving a system of equations via matrix methods

1. Convert system to matrix-vector equation
2. Augment the coefficients matrix with the vector of constants
3. Use Gaussian elimination to create a matrix with 1's on diagonals
4. Map the matrix back to the equations
5. Back-substitution to solve for variables
6. Check your work using the original system

# Reduced Row Echelon Form (RREF)

```python
import numpy as np
from sympy import Matrix

A = Matrix(np.random.randn(4,4))
B = Matrix(np.random.randn(4,3))

# compute RREF
rrefA = A.rref()
rrefB = B.rref()

print(np.array(rrefA[0]))
print(' ')
print(np.array(rrefB[0]))
```
