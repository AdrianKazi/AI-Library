# Dot products properties: associative and distributive

The dot product is distributive

$$a(b+c) = ab + ac $$

$$ a^T(b+c) = a^Tb + a^Tc $$

$$ \sum_{i=1}^n a_ib_i + \sum_{i=1}^n a_ic_i = \sum_{i=1}^n a_i(b_i + c_i)$$

The dot product is not associative

Skalars:

$$ a(b \cdot c) = (a\cdot b)c  $$

But:

$$ a^T(b^Tc) \neq (a^Tb)^Tc $$

```python
n = 10
a = np.random.randn(n)
b = np.random.randn(n)
c = np.random.randn(n)

res1 = np.dot(a, (b+c))
res2 = np.dot(a,b) + np.dot(a,c)

res1, res2
```

```python
# Associativee property

n = 5
a = np.random.randn(n)
b = np.random.randn(n)
c = np.random.randn(n)

res1 = np.dot(a, np.dot(b,c))
res2 = np.dot(np.dot(a,b), c)

res1, res2
```

# Dot prod: geometric perspective

$$  \alpha = a^T b = \sum_{i=1}^n a_i b_i = cos(\theta_{ab})\, ||a||\, ||b||  $$

$$v_1 \times v_2 = ||v_1|| \times ||v_2|| \times \cos(\theta)$$
$$\cos(\theta) = \frac{v_1 \times v_2}{||v_1||\times ||v_2||}$$
$$ \theta = \arccos \Big(\frac{v_1 \times v_2}{||v_1||\times ||v_2||}\Big)$$

```python
v1 = np.array([2,4,-3])
v2 = np.array([0,-3,-3])

theta = np.arccos((v1 @ v2) / (np.linalg.norm(v1) * np.linalg.norm(v2)))

fig = plt.subplots(subplot_kw = {'projection':'3d'})
plt.plot([0,v1[0]], [0,v1[1]], [0, v1[2]], 'r')
plt.plot([0,v2[0]], [0,v2[1]], [0, v2[2]], 'b')
plt.axis((-6,6,-6,6,-3,0))
plt.tight_layout()
plt.title(f'Angle between vectors: {theta:.5f}')
plt.show()
```

# Cross-product

$$Ax + By + Cz = 0$$
$$ Cz = -Ax - By $$
$$ z = \frac{-Ax-By}{C} $$

# Subspace

A vector subspace must

* be closed under addition and scalar multiplication
* contain the zero vector

$$ \forall v,w \in V;\;\;\forall \lambda,\alpha \in \mathbb{R};\;\; \lambda v + \alpha w \in V $$

If a vector $w$ is a scalar multiple of $v$ (so $ w = \lambda v $), then the set of all multiples of v is a subspace:

$$ \text{Span}(v) = \{ \lambda v \mid \lambda \in \mathbb{R}\} $$

```python
import numpy as np
import matplotlib.pyplot as plt

figure = plt.figure(figsize = (8,8))

dim = 2

# generate random vector
v1 = np.random.randn(dim)
plt.plot(v1[0], v1[1], 'ro', label='Original Vector')

lambdas = np.linspace(-10,10,100)
v1_span = np.array([lam * v1 for lam in lambdas])

plt.plot(v1_span[:,0], v1_span[:,1], label = 'Span')
plt.title(f'$\mathbb{{R}}^{2}$ Span of\n$v$: {v1}')
plt.grid()
plt.legend()
plt.axis('square')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

v1 = 10 * np.random.randn(3)
v2 = 10 * np.random.randn(3)

lambda1 = np.linspace(-10,10,30)
lambda2 = np.linspace(-10,10,30)

L1, L2 = np.meshgrid(lambda1, lambda2)

plane = L1[...,None] * v1 + L2[...,None] * v2

fig, ax = plt.subplots(subplot_kw = {'projection':'3d'}, figsize = (8,8))

ax.plot_surface(plane[:,:,0], plane[:,:,1], plane[:,:,2], alpha = .5, color = 'gold')

ax.quiver(0,0,0,v1[0],v1[1],v1[2], color = 'red', linewidth = 3)
ax.quiver(0,0,0,v2[0],v2[1],v2[2], color = 'blue', linewidth = 3)

plt.title(f'$\mathbb{{R}}^{3}$ Span of\n$v_1$: {v1}\n$v_2$:{v2}')
plt.show()
```

```python
import numpy as np
import plotly.graph_objects as go

# tworzymy figurę
fig = go.Figure()

# powierzchnia
fig.add_trace(go.Surface(
    x=plane[:, :, 0],
    y=plane[:, :, 1],
    z=plane[:, :, 2],
    opacity=0.6,
    colorscale="YlOrBr",
    showscale=False
))

# wektor v1
fig.add_trace(go.Cone(
    x=[0], y=[0], z=[0],
    u=[v1[0]], v=[v1[1]], w=[v1[2]],
    colorscale=[[0, 'red'], [1, 'red']],
    showscale=False,
    sizemode="absolute",
    sizeref=30
))

# wektor v2
fig.add_trace(go.Cone(
    x=[0], y=[0], z=[0],
    u=[v2[0]], v=[v2[1]], w=[v2[2]],
    colorscale=[[0, 'blue'], [1, 'blue']],
    showscale=False,
    sizemode="absolute",
    sizeref=30
))

# ustawienia osi
fig.update_layout(
    title=f"Span(v₁, v₂) in ℝ³<br>v₁ = {v1}<br>v₂ = {v2}",
    scene=dict(
        xaxis_title="X",
        yaxis_title="Y",
        zaxis_title="Z",
        aspectmode="cube"
    ),
    width=900,
    height=900
)

fig.show()
```

# Span

$$ span({v_1, ...,v_n}) = \alpha_1 v_1 + ... + \alpha_n v_n, \alpha \in \mathbb{R} $$

Span of set of vectors is all possible linear combinations of all the vectors in that set.

```python
import numpy as np
import matplotlib.pyplot as plt

S1 = np.array([1,1,0])
S2 = np.array([1,7,0])

# vectors v and w which are in span of set
v = np.array([1,2,0])
w = np.array([3,2,1])

# draw vectors
fig, ax = plt.subplots(subplot_kw = {'projection':'3d'})
ax.plot([0,S1[0]], [0, S1[1]], [.1, S1[2]+.1], 'r', linewidth = 3)
ax.plot([0,S2[0]], [0, S2[1]], [.1, S2[2]+.1], 'r', linewidth = 3)
ax.plot([0,v[0]], [0, v[1]], [.1, v[2]+.1], 'g', linewidth = 3)
ax.plot([0,w[0]], [0, w[1]], [.1, w[2]+.1], 'b', linewidth = 3)

# now draw plane
xx, yy = np.meshgrid(range(-15,16), range(-15,16))
cp = np.cross(S1,S2)
z1 = (-cp[0]*xx - cp[1]*yy)/cp[2]
ax.plot_surface(xx,yy,z1,alpha=.7)
plt.show()
```

# Linear Independence

It's a property of entire set of vectors, not just a single set.

Independent set of vectors is a set where no vector lies in the span of the other vectors.

Dependent set of vectors is a set where at least one vector can be created by a combination of the other vectors (not necessarily all of them).

# Basis

Basis - minimal set of vectors which:

* spans whole field (so we can do from it any vector)
* is linearly independent (so no vector is done from any other vector)

For example:

$$ \mathbb{R}^2 \rightarrow (1,0),(0,1)$$
$$ \mathbb{R}^3 \rightarrow (1,0,0),(0,1,0),(0,0,1)$$
