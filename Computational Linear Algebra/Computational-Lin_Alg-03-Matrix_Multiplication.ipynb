## Multiply by Layering

```python
m = 4
n = 6

A = np.random.randn(m,n)
B = np.random.randn(n,m)

# build the product matrix layer-wise
C1 = np.zeros((m,m))
for i in range(n):
  C1 += np.outer(A[:,i], B[i,:])

C2 = A@B
C2
```

## Order-of-operations on matrices

$$ (LIVE)^T = E^T V^T I^T L^T $$

```python
n = 7
L = np.random.randn(n,n)
I = np.random.randn(n,n)
V = np.random.randn(n,n)
E = np.random.randn(n,n)

# results of "forward" multiplication and then transpose
res1 = np.matrix.transpose(L@I@V@E)

# result of "flipped" multiplication of transposed matrices
res2 = E.T @ V.T @ I.T @ L.T

np.allclose(res1,res2)
```

```python
res1
```

```python
res2
```

## Matrix-Vector Multiplication

$$ A_{mxn}\;w_{nx1} = v_{mx1} $$
$$ w^T_{1xm}\;A_{mxn} = v_{1xn} $$

* If the vector is on the right, it has to be a column vector and result will be also a column vector
* If the vector is on the left, is has to be a row vector and result till be also a row vector

$$ Aw \rightarrow \text{weighted combinations of the columns of A} $$
$$ w^TA \rightarrow \text{weighted combinations of the rows of A} $$

```python
v_col = np.random.randn(5,1)
v_col
```

```python
v_row = np.random.randn(1, 5)
v_row
```

```python
res_col = matrix @ v_col
res_col, res_col.shape
```

```python
res_row = v_row @ matrix
res_row, res_row.shape
```

## 2D transformation matrices

$$ Av = \lambda v$$

Then $v$ is eigen-vector of matrix $A$ and $\lambda$ is corresponding eigen-value of matrix $A$

```python
import numpy as np
import matplotlib.pyplot as plt

# 2D input vector
v = np.array([3,-2])

# 2x2 transformation matrix
A = np.array([[1,-1],[2,1]])

# output vector is Av (convert v to column)
w = A @ v.T

# plot them
plt.plot([0,v[0]], [0,v[1]], label = 'v')
plt.plot([0,w[0]], [0,w[1]], label = 'Av')

plt.grid()
plt.axis((-6,6,-6,6))
plt.legend()
plt.title('Rotation + stretching')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import math

# 2D input vector
v = np.array([3,-2])

# 2x2 transformation matrix
th = 10*np.pi/24
A = np.array([[math.cos(th), -math.sin(th)], [math.sin(th), math.cos(th)]])

# output vector is Av (convert v to column)
w = A @ v.T

# plot them
plt.plot([0,w[0]], [0,w[1]], '-s',label = 'Av')
plt.plot([0,v[0]], [0,v[1]],label = 'v')

plt.grid()
plt.axis((-6,6,-6,6))
plt.legend()
plt.title('Pure Rotation')
plt.show()
```

## Pure and impure rotation matrices

```python
import numpy as np
import matplotlib.pyplot as plt
import math

# 2D input vector
v = np.array([3,-2])

thetas = np.linspace(0,2*np.pi,100)
vecmags = np.zeros((len(thetas), 2))

for i in range(len(thetas)):

  # 2x2 transformation matrix
  th = thetas[i]
  A1 = np.array([[2*math.cos(th), -math.sin(th)],
                 [math.sin(th), math.cos(th)]])
  A2 = np.array([[math.cos(th), -math.sin(th)],
                 [math.sin(th), math.cos(th)]])

  # compute vector magnitudes
  vecmags[i,0] = np.linalg.norm(A1@v.T)
  vecmags[i,1] = np.linalg.norm(A2@v.T)


plt.plot(thetas, vecmags, 'o-')
plt.xlabel('Rotation angle (rad.)')
plt.ylabel('Av magnitude')
plt.legend(['impure rotation', 'pure rotation'])
plt.show()
```

## Geometric transformations via matrix multiplications

```python
import matplotlib.pyplot as plt
import numpy as np
import math

coordinates = np.random.randn(2)
# transformation_matrix = np.eye(2)
transformation_matrix = np.random.randn(2,2)

fig = plt.figure(figsize = (8,8))

for i in range(100):
  theta = i*np.pi/45
  A = np.array([
      [math.cos(theta), -math.sin(theta)],
      [math.sin(theta), math.cos(theta)],
  ])

  w = coordinates @ A
  transformed_w = transformation_matrix @ w

  plt.plot(w[0], w[1], 'bo', markerfacecolor='w')
  plt.plot(transformed_w[0], transformed_w[1], 'go', alpha=.5)
plt.axis('square')
plt.show()
```

Symetric vs symetric matrices

Two symmetric matrices return symmetric matrix in addition and Hadamard multiplication operations. They don't return symmetric matrix in dot product.

```python
import numpy as np

A = np.random.randn(5,5)
B = np.random.randn(5,5)

A = (A + A.T) / 2
B = (B + B.T) / 2
```

```python
res1 = A + B
res1 == res1.T
```

```python
res2 = A @ B
res2 == res2.T
```

```python
res3 = A * B
res3 == res3.T
```

## Fourier transform via matrix multiplication

$$ F_{j,k} = w^m $$
$$ w = e^{-2\pi \sqrt{-1}/n} $$
$$ m = (j-1)(k-1) $$
$$ X = Fx $$

```python
np.sqrt(-1 + 0j)
```

```python
import numpy as np
import matplotlib.pyplot as plt

n = 52
F = np.zeros((n,n), dtype=complex)
w = np.exp((-2*np.pi*np.sqrt(-1 + 0j))/n)


for i in range(n):
  for j in range(n):
    m = i*j
    F[j,i] = np.power(w,m)


plt.plot(np.real(F[:,2]), 'k', label = 'col 2')
plt.plot(np.real(F[:,30]), 'y', label = 'col 30')
plt.grid()
plt.legend()
plt.show()
```

## Matrix Norms

There are 3 families of Matrix Norms

Similar to the vector norm:

* Square all individual matrix elements, add them tohether, take the square root.

$$ norm(A) = \sqrt{<A,A>_F} = \sqrt{tr(A^TA)} $$

```python
import numpy as np

# Create a matrix
A = np.array([[1,2,3], [4,5,6], [7,7,9]])

# optional orthogonal matrix to show that 2-norm is 1
Q, R = np.linalg.qr(A)
A, Q, R
```

```python
A, Q@R
```

```python
# Frobenius norm
norm_frob = np.linalg.norm(A, 'fro')
norm_frob
```

```python
norm_ind_2 = np.linalg.norm(A, 2)
norm_ind_2
```

```python
lamb = np.sqrt(np.max(np.linalg.eig(A.T@A)[0]))
lamb
```

```python
# scatten p-norm
p = 1
s = np.linalg.svd(A)[1] # get singular values
norm_schat = np.sum(s**p)**(1/p)
norm_schat
```
